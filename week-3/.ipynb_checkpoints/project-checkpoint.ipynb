```python
import tkinter as tk
from tkinter import messagebox
import csv

def load_employees(filename):
    """Load employee data from CSV into a list of dictionaries."""
    employees = []
    try:
        with open(filename, mode='r', encoding='utf-8') as file:
            reader = csv.DictReader(file)
            for row in reader:
                name = row.get('Name', row.get('name', '')).strip()
                dept = row.get('Department', row.get('department', '')).strip()
                if name and dept:
                    employees.append({'name': name, 'department': dept})
    except FileNotFoundError:
        messagebox.showerror("Error", f"File '{filename}' not found.")
        return None
    except Exception as e:
        messagebox.showerror("Error", f"Failed to read file: {e}")
        return None
    return employees

def show_department_members(employee, all_employees):
    """Create a new window listing other employees in the same department."""
    dept = employee['department']
    same_dept = [emp['name'] for emp in all_employees if emp['department'].lower() == dept.lower()]
    others = [name for name in same_dept if name.lower() != employee['name'].lower()]

    win = tk.Toplevel(root)
    win.title(f"Department: {dept}")
    win.geometry("400x300")

    tk.Label(win, text=f"Department: {dept}", font=('Arial', 14, 'bold')).pack(pady=10)
    tk.Label(win, text="Other members:", font=('Arial', 12)).pack(pady=5)

    if others:
        listbox = tk.Listbox(win, width=40, height=10)
        for name in others:
            listbox.insert(tk.END, name)
        listbox.pack(pady=10, expand=True, fill=tk.BOTH)
    else:
        tk.Label(win, text="No other members in this department.").pack(pady=10)

def verify():
    """Called when the 'Verify' button is clicked."""
    name = name_entry.get().strip()
    dept = dept_entry.get().strip()

    if not name or not dept:
        messagebox.showwarning("Input Error", "Please enter both name and department.")
        return

    if employees is None:
        messagebox.showerror("Error", "Employee data not available.")
        return

    found = None
    for emp in employees:
        if emp['name'].lower() == name.lower() and emp['department'].lower() == dept.lower():
            found = emp
            break

    if found:
        messagebox.showinfo("Welcome", f"Welcome {found['name']} from {found['department']}!")
        show_department_members(found, employees)
    else:
        messagebox.showinfo("Not Found", "Employee does not exist. Please check your details.")

root = tk.Tk()
root.title("GIG Logistics - Employee Verification")
root.geometry("400x200")

employees = load_employees("GIG-logistics.csv")

tk.Label(root, text="Name:", font=('Arial', 12)).pack(pady=5)
name_entry = tk.Entry(root, width=30)
name_entry.pack(pady=5)

tk.Label(root, text="Department:", font=('Arial', 12)).pack(pady=5)
dept_entry = tk.Entry(root, width=30)
dept_entry.pack(pady=5)

tk.Button(root, text="Verify", command=verify, bg="blue", fg="white").pack(pady=15)

if employees is None or len(employees) == 0:
    tk.Label(root, text="Warning: No employee data loaded. Check CSV file.", fg="red").pack()

root.mainloop()
```
